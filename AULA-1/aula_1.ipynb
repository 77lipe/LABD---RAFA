# Dimensões de Qualidade de Dados
- Completude
    * Ausência de dados  
- Consistência
    * Dado predominante em diferentes fontes
- Conformidade
    * Mesmo formato de Dado
- Integridade
    * Perder relação com fontes diferentes 
- Precisão
    * Informações que não correspondem a realidade 
- Atualidade
    * Tempo da informação
