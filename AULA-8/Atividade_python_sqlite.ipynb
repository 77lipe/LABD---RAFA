```python
import requests
from bs4 import BeautifulSoup
import sqlite3
import pandas as pd
```

# Parte 1: Configuração e Criação do Banco de Dados

---


Criação
do banco de dados e conexão: Crie um novo banco de dados SQLite
chamado empresa.db.
Conecte-se a ele e verifique se a conexão foi bem concluída.

```python
conexao = sqlite3.connect("empresa.db")
cursor = conexao.cursor()
```

Criação
de tabela

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS funcionarios(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    nome TEXT NOT NULL,
    cargo text not null,
    salario real,
    data_contratacao texto not null)
""")

conexao.commit()
```

```python
cursor.execute("PRAGMA table_info(funcionarios);")
cursor.fetchall()
```

# Parte 2: Manipulação de Dados (DML)

---

### Inserção de dados (INSERT):
 Insira pelo menos 5 registros na tabela funcionários com
dados fictícios.

```python
cursor.executemany("insert into funcionarios VALUES (?, ?, ?, ?, ?)", [
    (1,"Felipe", "Analista de Dados", 5000.00, "2022-01-15"),
    (2, "Alice", "Arquiteta", 7000.00, "2021-09-10"),
    (3, "Acsa", "Engenheira de Software", 6000.00, "2023-03-20"),
    (4, "Aline", "Cientista de Dados", 8000.00, "2022-11-05")
])

conexao.commit()
```

### Consulta
de dados (SELECT): Escreva um comando SELECT para exibir todos os registros da tabela funcionários.


```python
cursor.execute("select * from funcionarios;")
pd.DataFrame(cursor.fetchall(), columns=["id", "nome", "cargo", "salario", "data_contratacao"])
```

### Filtro - WHERE:

- Selecione apenas os funcionários com salário superior a R$ 5000.
- Selecione os funcionários que trabalham como 'Gerente'.


```python
cursor.execute("select * from funcionarios where salario > 5000;")
pd.DataFrame(cursor.fetchall(),columns=['id', 'nome', 'cargo', 'salario', 'data_contratacao'])
```

```python
cursor.execute("select * from funcionarios where cargo = 'Analista de Dados';")
pd.DataFrame(cursor.fetchall(), columns=['id', 'nome', 'cargo', 'salario', 'data_contratacao'])
```

### Atualização de dados (UPDATE):

```python
cursor.execute("update funcionarios set salario = salario * 1.1 where nome = 'Alice';")
conexao.commit()
```

### Exclusão de dados (DELETE):
 Exclui um dos funcionários que você inseriu.

```python
cursor.execute("delete from funcionarios where nome = 'Felipe';")
conexao.commit()
```

### Contagem de registros:
 Conte o número total de funcionários na tabela.

```python
cursor.execute("select count(*) from funcionarios;")
cursor.fetchall()
```
